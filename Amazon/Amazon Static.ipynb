```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
url="https://www.amazon.in/s?k=chair&crid=2RQRIV19W4KO8&sprefix=%2Caps%2C357&ref=nb_sb_ss_recent_1_0_recent"
```

```python
headers= {"user-agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36 viewport-width 887"}
```

```python
response = requests.get(url,headers= headers)
response.status_code
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
print(soup.prettify())
```

```python
parent_class="a-section a-spacing-base desktop-grid-content-view"
```

```python
products = soup.find_all("div", class_= parent_class)
```

```python
print(len(products))
```

```python
names = []
prices = []

products = soup.find_all("div", {"data-component-type": "s-search-result"})

for product in products:
    title_tag = product.find("h2", {"class": "a-size-base-plus a-spacing-none a-color-base a-text-normal"})
    name = title_tag.get_text(strip=True) if title_tag else "N/A"
    
    price_tag = product.find("span", {"class": "a-offscreen"})
    price = price_tag.get_text(strip=True) if price_tag else "N/A"
    
    names.append(name)
    prices.append(price)

df = pd.DataFrame({
    "Chair Name": names,
    "Price": prices
})

df
```
