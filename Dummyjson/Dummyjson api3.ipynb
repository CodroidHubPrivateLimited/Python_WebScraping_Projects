```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
url="https://dummyjson.com/products"
```

```python
headers = {"user-agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36"}
```

```python
response = requests.get(url,headers= headers)
```

```python
response.status_code
```

```python
data = response.json()
```

```python
main_data=data["products"]
```

```python
main_data[0].keys()
```

```python
id_list = []
title_list = []
price_list = []
discountPercentage =[]
rating = []
stock = []
for main in main_data:
    id_list.append(main["id"])
    title_list.append(main["title"])
    price_list.append(main["price"])
    discountPercentage.append(main["discountPercentage"])
    rating.append(main["rating"])
    stock.append(main["stock"])
df = pd.DataFrame({
    "id": id_list,
    "title": title_list,
    "price": price_list,
    "discountPercentage":discountPercentage,
    "rating":rating,
    "stock":stock
})
df
```
