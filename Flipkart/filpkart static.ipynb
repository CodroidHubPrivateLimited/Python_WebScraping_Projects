```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
url="https://www.flipkart.com/all/~cs-6ef68bc8d283b86730515a8f2c87ff23/pr?sid=0pm,fcn,821,a7x,2si&marketplace=FLIPKART&restrictLocale=true"
headres={"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36"}
```

```python
soup = BeautifulSoup(response.content,"html.parser")
```

```python
parent_class="RGLWAk"
products = soup.find_all("div", class_= parent_class)
print(len(products))
```

```python
names = []
prices = []
for product in products:
    name = product.find("a", class_="pIpigb").get_text()
    price = product.find("div",class_="hZ3P6w").get_text()
    names.append(name)
    prices.append(price)
   
df = pd.DataFrame({
    "Product": names,
    "Price": prices
})
df
```
