```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
url="https://automationexercise.com/brand_products/Polo"
```

```python
headers={"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36"}
```

```python
response = requests.get(url,headers= headers)
response.status_code
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
print(soup.prettify())
```

```python
parent_class="productinfo text-center"
```

```python
products = soup.find_all("div", class_= parent_class)
print(len(products))
```

```python
names = []
prices = []
for product in products:
    name = product.find("p").get_text()
    price = product.find("h2").get_text()
    names.append(name)
    prices.append(price)
df = pd.DataFrame({
    "Product": names,
    "Price": prices
})
df
```
