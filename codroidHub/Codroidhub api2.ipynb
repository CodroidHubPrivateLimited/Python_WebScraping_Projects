```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
url="https://puirf.codroidhub.com/api/api/v1/instruments/user/getHomeFacilities?upcomingLimit=2"
```

```python
headers = {"user-agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36"}
```

```python
response = requests.get(url,headers= headers)
```

```python
response.status_code
```

```python
data= response.json()
```

```python
main_data=data["data"]
```

```python
main_data.keys()
```

```python
saif_data = main_data["saif"]
```

```python
name_list = []
unique_id_list = []

for saif in saif_data:
    name_list.append(saif["name"])
    unique_id_list.append(saif["uniqueId"])
    
df = pd.DataFrame({
    "Name": name_list,
    "Unique_Id": unique_id_list
})
df
```

```python
cil_data = main_data["cil"]
```

```python
name_list = []
unique_id_list = []

for cil in cil_data:
    name_list.append(cil["name"])
    unique_id_list.append(cil["uniqueId"])
    
df = pd.DataFrame({
    "Name": name_list,
    "Unique_Id": unique_id_list
})
df
```

```python
upcoming_data = main_data["upcoming"]
```

```python
name_list = []
unique_id_list = []

for upcoming in upcoming_data:
    name_list.append(upcoming["name"])
    unique_id_list.append(upcoming["uniqueId"])
    
df = pd.DataFrame({
    "Name": name_list,
    "Unique_Id": unique_id_list
})
df
```
