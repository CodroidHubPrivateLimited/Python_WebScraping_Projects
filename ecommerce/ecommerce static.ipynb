<h1>Static Website Scraping - ecommerce Scrape</h1>
<h3>Data Extracted:  Name, Price</h3>
<h4>Tools Used: Requests, BeautifulSoup, Pandas</h4>


```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
url="https://www.scrapingcourse.com/ecommerce/"
```

```python
headers ={"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36"}
```

```python
response = requests.get(url,headers= headers)
response.status_code
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
print(soup.prettify())
```

```python
parent_class="woocommerce-LoopProduct-link woocommerce-loop-product__link"
```

```python
products = soup.find_all("a", class_= parent_class)
```

```python
print(len(products))
```

```python
import pandas as pd
name_list = []
price_list = []
for product in products:
    name_tag = product.find("h2")
    price_tag = product.find("span", class_="price")   
    if name_tag is not None and price_tag is not None:
        name = name_tag.get_text(strip=True)
        price = price_tag.get_text(strip=True)
        
        name_list.append(name)
        price_list.append(price)
       

df = pd.DataFrame({
    "Clothes_name": name_list,
    "Price": price_list
})

df
```
