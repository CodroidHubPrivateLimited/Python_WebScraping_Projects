```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time

```

```python
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
url = "https://www.meesho.com/search?q=saree"
driver.get(url)
time.sleep(5) 

products = driver.find_elements(By.CSS_SELECTOR, 'div[class*="ProductListItem__GridCol"]')
data_list = []

print(f"Found {len(products)} sarees!")

for item in products:
    name = item.find_element(By.TAG_NAME, "p").text
    price = item.find_element(By.TAG_NAME, "h5").text
    
    data_list.append({"Product Name": name, "Price": price})

driver.quit()

df = pd.DataFrame(data_list)
df
```
